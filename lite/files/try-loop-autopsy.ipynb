## 1. A stuck loop
`count` never changes, so the loop can never end.

```python
count = 1
while count <= 5:
    print(count)
```

## 2. A loop that jumps over its stop value
`x` goes 0, 3, 6, 9, 12 and never equals 10.

```python
x = 0
while x != 10:
    x += 3
```

## 3. A list that keeps growing
The loop adds to the list it is reading.

```python
nums = [1, 2]
for n in nums:
    nums.append(n)
```

## 4. A counter moving the wrong way

```python
i = 10
while i > 0:
    i += 1
```

## 5. Values going round in a cycle

```python
a = 0
while a < 5:
    a = (a + 1) % 3
```

## 6. Runaway recursion

```python
def f(n):
    return f(n)

f(3)
```

## 7. A correct loop is left alone

```python
total = 0
for i in range(1_000_000):
    total += i
print("done", total)
```

## 8. Settings
Change a setting, then put everything back.

```python
%autopsy config time_limit=120
```

```python
%autopsy reset
```

## 9. Run a cell without watching
`%%autopsy off` on the first line skips loop-autopsy for that cell.

```python
%%autopsy off
q = 0
while q < 1000:
    q += 1
print(q)
```
